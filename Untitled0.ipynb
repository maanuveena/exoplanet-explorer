```python
import pandas as pd

df = pd.read_csv(
    "PSCompPars_2026.10.17_23.57.54.csv",
    skiprows=20
)

print("Data loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python
print("Total exoplanets:", len(df))
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
# Count exoplanets by discovery method

method_counts = df["discoverymethod"].value_counts()

print("Exoplanets by discovery method:")
display(method_counts.head(10))
```

```python
import matplotlib.pyplot as plt

method_counts = df["discoverymethod"].value_counts().head(10)

method_counts.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Exoplanet Discovery Methods")
plt.xlabel("Discovery Method")
plt.ylabel("Number of Exoplanets")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
print(df.columns.tolist())
```

```python
import matplotlib.pyplot as plt

# Keep rows with planet size and orbital period available
planet_data = df[["pl_rade", "pl_orbper"]].dropna()

plt.figure(figsize=(8, 5))
plt.scatter(planet_data["pl_orbper"], planet_data["pl_rade"], alpha=0.5)

plt.xlabel("Orbital Period (days)")
plt.ylabel("Planet Radius (Earth radii)")
plt.title("Exoplanet Size vs Orbital Period")
plt.tight_layout()
plt.show()
```

```python
# Search for an exoplanet by name

planet_name = input("Enter an exoplanet name: ")

matches = df[
    df["pl_name"].str.contains(
        planet_name,
        case=False,
        na=False
    )
]

if not matches.empty:
    display(matches[["pl_name", "hostname", "discoverymethod",
                     "pl_rade", "pl_orbper"]].head(10))
else:
    print("No matching exoplanet found. Try another name.")
```

```python
# Show details of one selected exoplanet

planet_name = input("Enter the exact exoplanet name: ")

planet = df[df["pl_name"].str.lower() == planet_name.lower()]

if not planet.empty:
    row = planet.iloc[0]

    print("EXOPLANET DETAILS")
    print("-----------------")
    print("Planet name:", row["pl_name"])
    print("Host star:", row["hostname"])
    print("Discovery method:", row["discoverymethod"])
    print("Planet radius (Earth radii):", row["pl_rade"])
    print("Orbital period (days):", row["pl_orbper"])
else:
    print("Planet not found. Enter an exact name from your search results.")
```

```python
# Compare exoplanet size with Earth

planet_name = input("Enter an exoplanet name: ")

planet = df[df["pl_name"].str.lower() == planet_name.lower()]

if not planet.empty:
    radius = planet.iloc[0]["pl_rade"]

    if pd.notna(radius):
        print("Planet:", planet.iloc[0]["pl_name"])
        print("Earth radius = 1")
        print("Exoplanet radius:", radius, "Earth radii")

        if radius > 1:
            print(f"This planet's radius is {radius:.2f} times Earth's radius.")
        elif radius < 1:
            print(f"This planet's radius is {radius:.2f} times Earth's radius.")
        else:
            print("This planet has the same radius as Earth.")
    else:
        print("Radius data is not available for this planet.")
else:
    print("Planet not found. Enter an exact planet name.")
```

```python
import matplotlib.pyplot as plt

# Select planets with known radius
size_data = df[["pl_name", "pl_rade"]].dropna()

# Show 10 planets with the largest known radii
largest = size_data.nlargest(10, "pl_rade")

plt.figure(figsize=(10, 6))
plt.barh(largest["pl_name"], largest["pl_rade"])
plt.xlabel("Planet Radius (Earth radii)")
plt.ylabel("Exoplanet")
plt.title("10 Largest Exoplanets in Our Dataset")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
